```python
# Task 1: Dirty Data Cleanup

import pandas as pd

path = r"C:\Users\User\Downloads\dirty_cafe_sales.csv"
df = pd.read_csv(path)

# 1. Check missing values
print("Missing values before cleaning:\n", df.isnull().sum())

# 2. Remove low-quality rows (less than 3 real values)
df = df.dropna(thresh=3)
print("\nShape after quality filter:", df.shape)

# 3. Replace remaining missing values with 0
df = df.fillna(0)
print("\nMissing values after fill:\n", df.isnull().sum())

# 4. Check and remove duplicate rows
print("\nDuplicate rows found:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)

# 5. Final dataset
print("\nFinal cleaned dataset preview:\n")
df.head()
```
