# Data Cleaning in Pandas

```python
# df.isnull().sum() counts missing values per column
# df.isnull().sum().sum() counts total missing values in dataset
import pandas as pd
path = r"C:\Users\User\Downloads\Students Social Media Addiction.csv"
df = pd.read_csv(path)
print(df.isnull().sum())
print("Total missing values in dataset: ",df.isnull().sum().sum())
```

```python
# dropna() method is used to remove empty rows
# inside dropna we write inplace = True to modify the dataframe without creating its copy
import pandas as pd
path = r"C:\Users\User\Downloads\Students Social Media Addiction.csv"
df = pd.read_csv(path)
df.dropna(inplace=True)
```

```python
# if we use fillna method like this it will work on copy not the actual dataframe
import pandas as pd

data = {
    "calories": [210, 432, None],
    "duration": [20, 30, 40]
}

df = pd.DataFrame(data, index=["Day1", "Day2", "Day3"])
print(df.isnull())

df["calories"].fillna(130, inplace=True)
```

```python
# .fillna(value, inplace = True) method replaces all empty cells in the whole Data Frame with that given value
# in this way we are working on actual dataframe
import pandas as pd

data = {
    "calories": [210, 432, None],
    "duration": [20, 30, 40]
}

df = pd.DataFrame(data, index=["Day1", "Day2", "Day3"])
print(df.isnull())

df.fillna({"calories": 130}, inplace=True)
```

```python
# Correcting the Wrong Data by replacing the value
import pandas as pd
data = {
    "calories": [210, 432, 340],
    "duration": [20, 30, 40]
}

df = pd.DataFrame(data, index=["Day1", "Day2", "Day3"])

for x in df.index:
    if df.loc[x, 'duration'] > 30:
        df.loc[x, 'duration'] = 30
print(df)
```

```python
#.duplicated() method returns True for every row that is duplicate, otherwise false
#.drop_duplicates(inplace=True) method is used to remove duplicated rows from original df
import pandas as pd
data = {
    "calories": [210, 30, 30],
    "duration": [20, 30, 30]
}

df = pd.DataFrame(data, index=["Day1", "Day2", "Day3"])

print("Duplicated row:\n",df.duplicated())

df.drop_duplicates(inplace=True)
print("After removing duplicates:\n",df)
```
