```python
import pandas as pd
s1_employees = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Sara", "Ahmed", "John", "Maria", "Ali"],
    "Role": ["AI Engineer", "Data Analyst", "ML Ops", "AI Engineer", "Data Analyst"],
    "Salary": [95000, 72000, 88000, 105000, 65000],
    "Experience": [3,1,5,4,2]
}
df = pd.DataFrame(s1_employees)
# show only AI Engineers who have more than 3 years of experience
print("AI Engineers with more than 3 years of experience:\n",df[(df["Role"] == "AI Engineer") & (df["Experience"] > 3)])
# calculate the average salary grouped by Role
print("Average salary grouped by Role:\n",df.groupby("Role")["Salary"].mean())

# add a column level: "Senior" if experience >= 4 else "Junior"
exp = []
for x in df["Experience"]:
    if x >= 4:
        exp.append("Senior")
    else:
        exp.append("Junior")
df["Level"] = exp
df.head()
```

```python
import numpy as np
# create a 5*5 array containing random integers 10 and 100
arr = np.random.randint(10, 100, (5, 5))
mean_arr = np.mean(arr)
print("All the values greater than the mean are:")
print(arr[arr > mean_arr])
    
arr[arr < 50] = 0

print("\nMean of the array is:", mean_arr)
print("Modified array: ")
print(arr)
```

```python
import pandas as pd
s1_employees = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Sara", "Ahmed", "John", "Maria", "Ali"],
    "Role": ["AI Engineer", "Data Analyst", "ML Ops", "AI Engineer", "Data Analyst"],
    "Salary": [95000, 72000, 88000, 105000, 65000],
    "Experience": [3,1,5,4,2]
}
df = pd.DataFrame(s1_employees)
df.loc[2, "Salary"] = None
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Salary"].sort_values(ascending=False)
```
